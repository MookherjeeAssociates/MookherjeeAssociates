```python
import pandas as pd

def calculate_financial_ratios(data):
    """
    Calculates key financial ratios for MSME health analysis.
    """
    print("--- Mookherjee Associates: Financial Ratio Output ---\n")

    # 1. Current Ratio (Liquidity)
    current_ratio = data['Current Assets'] / data['Current Liabilities']
    print(f"Current Ratio: {current_ratio:.2f}")
    if current_ratio >= 1.33:
         print(" Status: Healthy (Meets standard bank lending norms)")
    else:
         print(" Status: Weak (May face liquidity issues)")

    # 2. Debt-to-Equity Ratio (Leverage)
    debt_equity = data['Total Outside Liabilities'] / data['Tangible Net Worth']
    print(f"\nDebt-to-Equity Ratio: {debt_equity:.2f}")
    if debt_equity <= 2.0:
        print(" Status: Optimal (Comfortable leverage)")
    else:
        print(" Status: High Risk (Over-leveraged)")

    # 3. Inventory Turnover Ratio (Efficiency)
    inv_turnover = data['Cost of Goods Sold (COGS)'] / data['Average Inventory']
    print(f"\nInventory Turnover Ratio: {inv_turnover:.2f} times")

    print("\n---------------------------------------------------")
    print("Note: This is a programmatic simulation. Final CMA data")
    print("requires manual adjustments for statutory compliance.")

# Dummy Data representing an MSME's financials (in INR)
msme_financials = {
    'Current Assets': 2500000,
    'Current Liabilities': 1750000,
    'Total Outside Liabilities': 4500000,
    'Tangible Net Worth': 3000000,
    'Cost of Goods Sold (COGS)': 8500000,
    'Average Inventory': 1200000
}

# Execute the function
calculate_financial_ratios(msme_financials)
```
